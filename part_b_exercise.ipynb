```python
def gc_content_loop(sequence):
    gc = 0

    for base in sequence:
        if base == "G" or base == "C":
            gc += 1

    return gc / len(sequence)
```

```python
import numpy as np

def gc_content_numpy(sequence):
    bases = np.array(list(sequence))
    return np.mean((bases == "G") | (bases == "C"))
```

```python
sequence = "GATCGGTCA"

result1 = gc_content_loop(sequence)
result2 = gc_content_numpy(sequence)

print(result1)
print(result2)
```
